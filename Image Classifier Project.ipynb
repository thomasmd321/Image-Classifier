````python
# TODO: Build and train your network
# (train.py has the same logic plus checkpoints on disk, resuming, mixed precision and more options)
ARCHS = {'densenet121': (1024, 'classifier', ['features.denseblock4', 'features.norm5']),
         'alexnet': (9216, 'classifier', ['features.10']),
         'vgg16': (25088, 'classifier', ['features.28']),
         'resnet50': (2048, 'fc', ['layer4'])}


def build_model(arch='densenet121', hidden_units=(512, 90, 80), num_classes=102, dropout=0.5, pretrained=True):
    in_features, head, _ = ARCHS[arch]
    try:
        model = getattr(models, arch)(weights='DEFAULT' if pretrained else None)
    except TypeError:
        # Older torchvision versions (< 0.13) only understand pretrained=
        model = getattr(models, arch)(pretrained=pretrained)
    # Freeze the pretrained feature parameters so only the new classifier is trained
    for param in model.parameters():
        param.requires_grad = False
    # Dropout, then one Linear + ReLU per hidden layer, then the output layer
    sizes = [in_features] + list(hidden_units)
    layers = [('dropout', nn.Dropout(dropout))]
    for i in range(1, len(sizes)):
        name = 'inputs' if i == 1 else 'hidden_layer{}'.format(i - 1)
        layers += [(name, nn.Linear(sizes[i - 1], sizes[i])), ('relu{}'.format(i), nn.ReLU())]
    layers += [('hidden_layer{}'.format(len(sizes) - 1), nn.Linear(sizes[-1], num_classes)),
               ('output', nn.LogSoftmax(dim=1))]
    setattr(model, head, nn.Sequential(OrderedDict(layers)))
    model.arch = arch
    return model


def unfreeze_last_block(model):
    params = []
    for name in ARCHS[model.arch][2]:
        for param in model.get_submodule(name).parameters():
            param.requires_grad = True
            params.append(param)
    return params


def evaluate(model, loader, criterion):
    model.eval()
    total_loss, correct, total = 0, 0, 0
    with torch.no_grad():
        for inputs, labels in loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            total_loss += criterion(outputs, labels).item()
            correct += (outputs.argmax(dim=1) == labels).sum().item()
            total += labels.size(0)
    model.train()
    return total_loss / len(loader), correct / total


def train_phase(model, params, lr, first_epoch, last_epoch, phase, patience=5):
    """ Trains `params` for epochs [first_epoch, last_epoch), keeping the best weights (by validation accuracy)
        in best_state, reducing the LR when validation loss stalls, and stopping early after `patience`
        epochs without improvement. """
    global best_state, best_accuracy
    optimizer = optim.Adam(params, lr=lr)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.1, patience=2)
    without_improvement = 0
    model.train()
    for e in range(first_epoch, last_epoch):
        epoch_loss = 0
        for inputs, labels in dataloaders['train']:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            loss = train_criterion(model(inputs), labels)
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()

        valid_loss, accuracy = evaluate(model, dataloaders['valid'], eval_criterion)
        history.append({'epoch': e + 1, 'phase': phase, 'train_loss': epoch_loss / len(dataloaders['train']),
                        'valid_loss': valid_loss, 'valid_accuracy': accuracy})
        scheduler.step(valid_loss)
        improved = best_accuracy is None or accuracy > best_accuracy
        if improved:
            best_accuracy, best_state = accuracy, copy.deepcopy(model.state_dict())
            without_improvement = 0
        else:
            without_improvement += 1
        print('Epoch: {}/{}... '.format(e + 1, total_epochs),
              'Training Loss: {:.4f}'.format(history[-1]['train_loss']),
              'Validation Loss: {:.4f}'.format(valid_loss),
              'Accuracy: {:.4f}{}'.format(accuracy, ' (best)' if improved else ''))
        if without_improvement >= patience:
            print('No improvement for {} epochs; stopping this phase early.'.format(patience))
            break


arch, hidden_units, dropout = 'densenet121', [512, 90, 80], 0.5
epochs, learning_rate = 10, 0.001            # phase 1: train the new classifier
finetune_epochs, finetune_lr = 5, 1e-4       # phase 2: also fine-tune the last backbone block
total_epochs = epochs + finetune_epochs
num_classes = len(image_datasets['train'].classes)

model = build_model(arch, hidden_units, num_classes, dropout)
model.to(device)
# Label smoothing on the training loss only. The model outputs log-probabilities; CrossEntropyLoss
# re-applies log_softmax, which leaves them unchanged, so this is NLLLoss plus label smoothing.
train_criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
eval_criterion = nn.NLLLoss()
history, best_state, best_accuracy = [], None, None

start_time = time.time()
train_phase(model, getattr(model, ARCHS[arch][1]).parameters(), learning_rate, 0, epochs, 'head')
if finetune_epochs:
    print('Fine-tuning the last {} block from the best classifier ...'.format(arch))
    model.load_state_dict(best_state)
    params = unfreeze_last_block(model) + list(getattr(model, ARCHS[arch][1]).parameters())
    train_phase(model, params, finetune_lr, epochs, total_epochs, 'finetune')

# Keep the best model found in either phase
model.load_state_dict(best_state)
elapsed = time.time() - start_time
print('Training complete in {:.0f}m {:.0f}s; best validation accuracy {:.4f}'.format(
    elapsed // 60, elapsed % 60, best_accuracy))
````

````python
# TODO: Write a function that loads a checkpoint and rebuilds the model
def load_checkpoint(path):
    checkpoint = torch.load(path, map_location=device)
    # Older checkpoints only stored the first hidden layer size
    hidden = checkpoint.get('hidden_layers') or [checkpoint['hidden_layer1'], 90, 80]
    model = build_model(checkpoint['structure'], hidden, checkpoint['num_classes'], checkpoint['dropout'],
                        pretrained=False)
    model.load_state_dict(checkpoint['state_dict'])
    model.class_to_idx = checkpoint['class_to_idx']
    model.to(device)
    model.eval()
    return model

model = load_checkpoint('check_point.pt')
````

## Image Preprocessing

You'll want to use `PIL` to load the image ([documentation](https://pillow.readthedocs.io/en/latest/reference/Image.html)). It's best to write a function that preprocesses the image so it can be used as input for the model. This function should process the images in the same manner used for training. 

First, resize the images where the shortest side is 256 pixels, keeping the aspect ratio. This can be done with the [`thumbnail`](http://pillow.readthedocs.io/en/3.1.x/reference/Image.html#PIL.Image.Image.thumbnail) or [`resize`](http://pillow.readthedocs.io/en/3.1.x/reference/Image.html#PIL.Image.Image.thumbnail) methods. Then you'll need to crop out the center 224x224 portion of the image.

Color channels of images are typically encoded as integers 0-255, but the model expected floats 0-1. You'll need to convert the values. It's easiest with a Numpy array, which you can get from a PIL image like so `np_image = np.array(pil_image)`.

As before, the network expects the images to be normalized in a specific way. For the means, it's `[0.485, 0.456, 0.406]` and for the standard deviations `[0.229, 0.224, 0.225]`. You'll want to subtract the means from each color channel, then divide by the standard deviation. 

And finally, PyTorch expects the color channel to be the first dimension but it's the third dimension in the PIL image and Numpy array. You can reorder dimensions using [`ndarray.transpose`](https://docs.scipy.org/doc/numpy-1.13.0/reference/generated/numpy.ndarray.transpose.html). The color channel needs to be first and retain the order of the other two dimensions.

````python
def process_image(image):
    ''' Scales, crops, and normalizes a PIL image for a PyTorch model,
        returns an Numpy array
    '''
    
    # TODO: Process a PIL image for use in a PyTorch model
    img = Image.open(image).convert('RGB')
    # Resize so the shortest side is 256 pixels, keeping the aspect ratio
    width, height = img.size
    if width < height:
        img = img.resize((256, int(256 * height / width)))
    else:
        img = img.resize((int(256 * width / height), 256))
    # Crop out the center 224x224 portion
    width, height = img.size
    left, top = (width - 224) // 2, (height - 224) // 2
    img = img.crop((left, top, left + 224, top + 224))
    # Scale to 0-1, normalize, and move the color channel to the first dimension
    np_image = np.array(img) / 255
    np_image = (np_image - np.array(norm_mean)) / np.array(norm_std)
    return np_image.transpose((2, 0, 1))
````

To check your work, the function below converts a PyTorch tensor and displays it in the notebook. If your `process_image` function works, running the output through this function should return the original image (except for the cropped out portions).

````python
def imshow(image, ax=None, title=None):
    """Imshow for Tensor."""
    if ax is None:
        fig, ax = plt.subplots()
    
    # PyTorch tensors assume the color channel is the first dimension
    # but matplotlib assumes is the third dimension
    image = image.numpy().transpose((1, 2, 0))
    
    # Undo preprocessing
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    image = std * image + mean
    
    # Image needs to be clipped between 0 and 1 or it looks like noise when displayed
    image = np.clip(image, 0, 1)
    
    ax.imshow(image)
    
    return ax
````

## Class Prediction

Once you can get images in the correct format, it's time to write a function for making predictions with your model. A common practice is to predict the top 5 or so (usually called top-$K$) most probable classes. You'll want to calculate the class probabilities then find the $K$ largest values.

To get the top $K$ largest values in a tensor use [`x.topk(k)`](http://pytorch.org/docs/master/torch.html#torch.topk). This method returns both the highest `k` probabilities and the indices of those probabilities corresponding to the classes. You need to convert from these indices to the actual class labels using `class_to_idx` which hopefully you added to the model or from an `ImageFolder` you used to load the data ([see here](#Save-the-checkpoint)). Make sure to invert the dictionary so you get a mapping from index to class as well.

Again, this method should take a path to an image and a model checkpoint, then return the probabilities and classes.

```python
probs, classes = predict(image_path, model)
print(probs)
print(classes)
> [ 0.01558163  0.01541934  0.01452626  0.01443549  0.01407339]
> ['70', '3', '45', '62', '55']
```

````python
def predict(image_path, model, topk=5):
    ''' Predict the class (or classes) of an image using a trained deep learning model.
    '''
    
    # TODO: Implement the code to predict the class from an image file
    model.to(device)
    model.eval()
    img = torch.from_numpy(process_image(image_path)).float().unsqueeze(0).to(device)
    with torch.no_grad():
        output = model(img)
    # The model outputs log-probabilities, so exp() gives the probabilities
    probs, indices = torch.exp(output).topk(topk)
    idx_to_class = {val: key for key, val in model.class_to_idx.items()}
    classes = [idx_to_class[index] for index in indices[0].tolist()]
    return probs[0].tolist(), classes

image_path = test_dir + '/10/image_07104.jpg'
probs, classes = predict(image_path, model)
print(probs)
print(classes)
````
